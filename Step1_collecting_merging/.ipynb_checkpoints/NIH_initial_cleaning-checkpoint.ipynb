### first import all the NIH datasets.

```python
import os
import csv
import pandas as pd

# Folder containing NIH CSVs
in_folder = "NIH/New"

# List of files to load
files = [
    "chitosan.csv",
    "hydrogel.csv",
    "PCL.csv",
    "PEG.csv",
    "PLA.csv",
    "PLGA.csv",
    "poly(lactic-co-glycolic acid).csv",
    "polycaprolactone.csv",
    "polyethylene_glycol.csv",
    "polylactic_acid.csv",
    "polymeric_micelle.csv"
]

nih_dfs = {}

for fname in files:
    fpath = os.path.join(in_folder, fname)
    with open(fpath, "r", encoding="utf-8", errors="ignore") as f:
        reader = csv.reader(f)
        rows = list(reader)

    # header row is at index 7, data starts from index 8
    df = pd.DataFrame(rows[8:], columns=rows[7])

    # Add Polymer column (based on file name without .csv)
    polymer_name = os.path.splitext(fname)[0]
    df["Polymer"] = polymer_name

    nih_dfs[polymer_name] = df
    print(f"{fname} → {df.shape[0]} rows, {df.shape[1]} columns")

# Example: preview chitosan
nih_dfs["chitosan"].head()
```

```python
# Step 2: Clean & merge NIH data

cleaned_dfs = []

for name, df in nih_dfs.items():
    # Standardize column names
    df.columns = [c.strip().lower().replace(" ", "_").replace("/", "_") for c in df.columns]

    # Deduplicate within each polymer by application_id
    if "application_id" in df.columns:
        df = df.drop_duplicates(subset=["application_id"])

    cleaned_dfs.append(df)

# Merge all into one big dataframe
nih_merged = pd.concat(cleaned_dfs, ignore_index=True)

# Deduplicate globally
if "application_id" in nih_merged.columns:
    nih_merged = nih_merged.drop_duplicates(subset=["application_id"])

                           
```

```python
# Step 3: Normalize polymer names
polymer_mapping = {
    "PCL": "polycaprolactone",
    "PLA": "polylactic acid",
    "polylactic_acid": "polylactic acid",
    "polyethylene_glycol": "PEG",
    "poly(lactic-co-glycolic acid)": "PLGA"
}

nih_merged["Polymer"] = nih_merged["Polymer"].replace(polymer_mapping)

# Step 4: Check duplicates
if "application_id" in nih_merged.columns:
    # Count before deduplication
    before_count = nih_merged.shape[0]

    # Find duplicated rows
    duplicates = nih_merged[nih_merged.duplicated(subset=["application_id"], keep=False)]
    
    if not duplicates.empty:
        print(f"⚠️ Found {duplicates.shape[0]} duplicate rows (by application_id):")
        print(duplicates[["application_id", "project_title", "Polymer"]].head(20))
    else:
        print("✅ No duplicate application_id found.")

    # Deduplicate
    nih_merged = nih_merged.drop_duplicates(subset=["application_id"])

    # Count after deduplication
    after_count = nih_merged.shape[0]
    print(f"\nDeduplicated NIH dataset: {before_count} → {after_count} rows")
else:
    print("⚠️ application_id column not found in dataset — cannot check duplicates.")

# Step 5: Verify unique polymers after normalization
print("\nUnique polymers after normalization:")
print(sorted(nih_merged["Polymer"].unique()))

# Step 6: Save final dataset
nih_merged.to_csv("NIH/New/nih_polymer_drug_delivery_cleaned.csv", index=False)
print("\n✅ Final NIH dataset saved as nih_polymer_drug_delivery_cleaned.csv")
```
