```python
import re
import os
from typing import Tuple
import pandas as pd

# ensure output folder exists
os.makedirs("Patent", exist_ok=True)

# -----------------------------
# 1) Config: delivery REGEX patterns & polymer keyword map
# -----------------------------
# Regex patterns catch plurals + hyphenation + common variants (all case-insensitive via lowercasing)
DELIVERY_PATTERNS = [
    r"\bdrug[-\s]?delivery\b",
    r"\bdrug[-\s]?delivery\s+system(s)?\b",
    r"\bcontrolled[-\s]?release\b",
    r"\bsustained[-\s]?release\b",
    r"\btargeted[-\s]?delivery\b",
    r"\bdrug[-\s]?carrier(s)?\b",
    r"\btherapeutic[-\s]?delivery\b",
    # broadened nano/carrier language
    r"\bnano[-\s]?particle(s)?\b",
    r"\bnano[-\s]?carrier(s)?\b",
    r"\bnano[-\s]?particulate(s)?\b",
    # very common phrasing in abstracts
    r"\bdrug\s+release\b",
    r"\brelease\s+profile(s)?\b"
]

POLYMER_TERMS = {
    "chitosan": ["chitosan"],
    "hydrogel": ["hydrogel", "hydrogels"],
    "PCL": ["pcl", "polycaprolactone"],
    "PEG": ["peg", "polyethylene glycol"],
    "PLA": ["pla", "polylactic acid", "poly(lactic acid)"],
    "PLGA": ["plga", "poly(lactic-co-glycolic acid)", "poly lactic co glycolic acid", "poly(lactide-co-glycolide)"],
    "poly_lactic_co_glycolic_acid": ["plga", "poly(lactic-co-glycolic acid)", "poly lactic co glycolic acid", "poly(lactide-co-glycolide)"],
    "polycaprolactone": ["polycaprolactone", "pcl"],
    "polyethylene_glycol": ["polyethylene glycol", "peg"],
    "polylactic_acid": ["polylactic acid", "pla", "poly(lactic acid)"],
    "polymeric_micelle": ["polymeric micelle", "polymeric micelles"]
}

# Columns we care most about (keep these first if present)
PRIMARY_COLS = [
    "#","Jurisdiction","Kind","Display Key","Lens ID",
    "Publication Date","Publication Year","Application Number","Application Date",
    "Title","Abstract","Claims","CPC Classifications","IPCR Classifications",
    "Applicants/Assignees","Inventors","Legal Status"
]

# -----------------------------
# 2) Helpers
# -----------------------------
def _to_str(x): 
    return "" if pd.isna(x) else str(x)

def _norm_text(s: str) -> str:
    s = _to_str(s)
    s = re.sub(r"\s+", " ", s).strip().lower()
    return s

def _tac_text(row) -> str:
    return " ".join([
        _to_str(row.get("Title","")),
        _to_str(row.get("Abstract","")),
        _to_str(row.get("Claims",""))
    ]).lower()

def _any_regex_match(patterns, text: str) -> bool:
    return any(re.search(p, text) for p in patterns)

def row_has_polymer_and_delivery(row, polymer_keywords_lower, delivery_patterns) -> bool:
    txt = _tac_text(row)  # already lowercased
    has_polymer = any(pk in txt for pk in polymer_keywords_lower)
    has_delivery = _any_regex_match(delivery_patterns, txt)
    return has_polymer and has_delivery

# -----------------------------
# 3) Clean one dataframe
# -----------------------------
def clean_single(df: pd.DataFrame, source_name: str) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """Return (kept_df, dropped_df) for one polymer source."""
    df = df.copy()

    # Reorder: put primary cols first but keep all others
    keep_first = [c for c in PRIMARY_COLS if c in df.columns]
    rest = [c for c in df.columns if c not in keep_first]
    df = df[keep_first + rest]

    # Add Source column
    df["Source"] = source_name

    # Parse dates if present
    for c in ["Publication Date","Application Date"]:
        if c in df.columns:
            df[c] = pd.to_datetime(df[c], errors="coerce")

    if "Publication Year" in df.columns:
        df["Publication Year"] = pd.to_numeric(df["Publication Year"], errors="coerce").astype("Int64")

    # Normalize TAC fields
    for c in ["Title","Abstract","Claims"]:
        if c in df.columns:
            df[c] = df[c].map(_norm_text)

    # Drop rows with all TAC empty
    if all(c in df.columns for c in ["Title","Abstract","Claims"]):
        df = df[~((df["Title"]=="") & (df["Abstract"]=="") & (df["Claims"]==""))]

    # De-dupe within source
    if "Lens ID" in df.columns:
        df = df.drop_duplicates(subset=["Lens ID"])
    # Fallback dedupe: Title + Publication Year
    fallback_cols = [c for c in ["Title","Publication Year"] if c in df.columns]
    if fallback_cols:
        df = df.drop_duplicates(subset=fallback_cols)

    # Polymer-specific validation (guard if source missing in map)
    polymer_keys = [k.lower() for k in POLYMER_TERMS.get(source_name, [source_name.lower()])]

    mask_keep = df.apply(
        row_has_polymer_and_delivery, axis=1,
        polymer_keywords_lower=polymer_keys,
        delivery_patterns=DELIVERY_PATTERNS
    )

    kept = df[mask_keep].copy()
    dropped = df[~mask_keep].copy()
    return kept, dropped

# -----------------------------
# 4) Apply to all loaded dataframes
# (requires you already defined chitosan_df, PEG_df, etc.)
# -----------------------------
name_to_df = {
    "chitosan": chitosan_df,
    "hydrogel": hydrogel_df,
    "PCL": PCL_df,
    "PEG": PEG_df,
    "PLA": PLA_df,
    "PLGA": PLGA_df,
    "poly_lactic_co_glycolic_acid": poly_lactic_co_glycolic_acid_df,
    "polycaprolactone": polycaprolactone_df,
    "polyethylene_glycol": polyethylene_glycol_df,
    "polylactic_acid": polylactic_acid_df,
    "polymeric_micelle": polymeric_micelle_df
}

kept_all, dropped_all, summary_rows = [], [], []

for name, df in name_to_df.items():
    start_n = len(df)
    kept, dropped = clean_single(df, name)
    kept_n, drop_n = len(kept), len(dropped)
    kept_all.append(kept)
    dropped_all.append(dropped)
    summary_rows.append({
        "Source": name,
        "Start": start_n,
        "Kept": kept_n,
        "Dropped": drop_n,
        "Kept_%": round(100 * kept_n / start_n, 2) if start_n else 0.0
    })
    print(f"{name}: start {start_n} → kept {kept_n}, dropped {drop_n}")

# -----------------------------
# 5) Merge & global de-dupe across sources
# -----------------------------
clean_df = pd.concat(kept_all, ignore_index=True)
if "Lens ID" in clean_df.columns:
    clean_df = clean_df.drop_duplicates(subset=["Lens ID"])
fallback_cols = [c for c in ["Title","Publication Year"] if c in clean_df.columns]
if fallback_cols:
    clean_df = clean_df.drop_duplicates(subset=fallback_cols)

dropped_df = pd.concat(dropped_all, ignore_index=True)

# -----------------------------
# 6) Save outputs
# -----------------------------
clean_df.to_csv("Patent/all_patents_clean.csv", index=False)
dropped_df.to_csv("Patent/all_patents_dropped.csv", index=False)

summary = pd.DataFrame(summary_rows).sort_values(by="Source")
#summary.to_csv("Patent/cleaning_summary.csv", index=False)

print("\n=== Cleaning Summary ===")
print(summary.to_string(index=False))
print(f"\nTOTAL kept: {len(clean_df)}")
print(f"TOTAL dropped: {len(dropped_df)}")
print("Saved: Patent/all_patents_clean.csv, Patent/all_patents_dropped.csv, Patent/cleaning_summary.csv")
```

```python
# Step 7: Normalize polymer names (same as NIH)
polymer_mapping = {
    "PCL": "polycaprolactone",
    "PLA": "polylactic acid",
    "polylactic_acid": "polylactic acid",
    "polyethylene_glycol": "PEG",
    "poly_lactic_co_glycolic_acid": "PLGA"
}

if "Source" in clean_df.columns:
    clean_df["Polymer"] = clean_df["Source"].replace(polymer_mapping)
else:
    print("⚠️ No Source column found in patent data.")

# Step 8: Check unique polymers
print("\nUnique polymers after normalization (Patents):")
print(sorted(clean_df["Polymer"].unique()))

# Step 9: Save normalized clean dataset
clean_df.to_csv("Patent/all_patents_cleaned.csv", index=False)
print("\n✅ Final Patent dataset saved as Patent/all_patents_cleaned.csv")

```
